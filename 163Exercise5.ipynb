```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import pandas as pd
```

```python
# Normal Distribution
mean = 50
std = 10
x = 60

# Probability density value
pdf = stats.norm.pdf(x, loc=mean, scale=std)

# Cumulative probability
cdf = stats.norm.cdf(x, loc=mean, scale=std)

# Mean, variance and standard deviation
normal_mean = stats.norm.mean(loc=mean, scale=std)
normal_variance = stats.norm.var(loc=mean, scale=std)
normal_std = stats.norm.std(loc=mean, scale=std)

print("Normal Distribution")
print("PDF at X =", x, ":", round(pdf, 4))
print("CDF at X =", x, ":", round(cdf, 4))
print("Mean :", normal_mean)
print("Variance :", normal_variance)
print("Standard Deviation :", normal_std)
```

```python
x_values = np.linspace(20, 80, 100)

y_values = stats.norm.pdf(
    x_values,
    loc=mean,
    scale=std
)

plt.plot(x_values, y_values)
plt.title("Normal Distribution")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid()
plt.show()
```

```python
# Binomial Distribution
n = 10
p = 0.5
k = 5

# Probability of exactly k successes
binomial_pmf = stats.binom.pmf(k, n, p)

# Cumulative probability P(X <= k)
binomial_cdf = stats.binom.cdf(k, n, p)

# Mean, variance and standard deviation
binomial_mean = stats.binom.mean(n, p)
binomial_variance = stats.binom.var(n, p)
binomial_std = stats.binom.std(n, p)

print("Binomial Distribution")
print("P(X =", k, ") :", round(binomial_pmf, 4))
print("P(X <=", k, ") :", round(binomial_cdf, 4))
print("Mean :", binomial_mean)
print("Variance :", binomial_variance)
print("Standard Deviation :", round(binomial_std, 4))
```

```python
x_values = np.arange(0, n + 1)

y_values = stats.binom.pmf(x_values, n, p)

plt.bar(x_values, y_values)
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.grid(axis='y')
plt.show()
```

```python
# Poisson Distribution
lam = 4
k = 3

# Probability of exactly k events
poisson_pmf = stats.poisson.pmf(k, mu=lam)

# Cumulative probability P(X <= k)
poisson_cdf = stats.poisson.cdf(k, mu=lam)

# Mean, variance and standard deviation
poisson_mean = stats.poisson.mean(mu=lam)
poisson_variance = stats.poisson.var(mu=lam)
poisson_std = stats.poisson.std(mu=lam)

print("Poisson Distribution")
print("P(X =", k, ") :", round(poisson_pmf, 4))
print("P(X <=", k, ") :", round(poisson_cdf, 4))
print("Mean :", poisson_mean)
print("Variance :", poisson_variance)
print("Standard Deviation :", round(poisson_std, 4))
```

```python
x_values = np.arange(0, 12)

y_values = stats.poisson.pmf(x_values, mu=lam)

plt.bar(x_values, y_values)
plt.title("Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.grid(axis='y')
plt.show()
```

```python
print("STATISTICAL SUMMARY")
print("-" * 50)

print("\nNormal Distribution")
print("Mean =", normal_mean)
print("Variance =", normal_variance)
print("Standard Deviation =", normal_std)

print("\nBinomial Distribution")
print("Mean =", binomial_mean)
print("Variance =", binomial_variance)
print("Standard Deviation =", round(binomial_std, 4))

print("\nPoisson Distribution")
print("Mean =", poisson_mean)
print("Variance =", poisson_variance)
print("Standard Deviation =", poisson_std)
```

```python
import pandas as pd
summary = {
    'Distribution': ['Normal', 'Binomial', 'Poisson'],
    'Mean': [normal_mean, binomial_mean, poisson_mean],
    'Variance': [normal_variance, binomial_variance, poisson_variance],
    'Standard Deviation': [normal_std, binomial_std, poisson_std]
}

summary_df = pd.DataFrame(summary)

print(summary_df.round(4))
```
