```python
import numpy as np

# 1. Array Creation
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Original Array:")
print(data)

# 2. Reshaping
reshaped = data.reshape(1, 9)

print("\nReshaped Array:")
print(reshaped)

# 3. Indexing
print("\nElement at row 2, column 3:")
print(data[1, 2])

# 4. Slicing
print("\nFirst two rows:")
print(data[:2])

# 5. Aggregation
print("\nSum:", np.sum(data))
print("Mean:", np.mean(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))

# 6. Min-Max Normalization
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))

print("\nNormalized Array:")
print(np.round(normalized, 2))

# 7. Standardization
standardized = (data - np.mean(data)) / np.std(data)

print("\nStandardized Array:")
print(np.round(standardized, 2))

# 8. Boolean Masking
mask = data > 50

print("\nBoolean Mask:")
print(mask)

# 9. Boolean Filtering
filtered = data[data > 50]

print("\nValues Greater Than 50:")
print(filtered)

# 10. Handling Missing Values
data_with_nan = np.array([10, 20, np.nan, 40, 50])

print("\nArray With Missing Value:")
print(data_with_nan)

mean_value = np.nanmean(data_with_nan)
data_with_nan[np.isnan(data_with_nan)] = mean_value

print("After Handling Missing Value:")
print(data_with_nan)

# 11. Sorting
unsorted = np.array([50, 20, 90, 10, 40])

print("\nSorted Array:")
print(np.sort(unsorted))

# 12. Concatenation
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

concatenated = np.concatenate((array1, array2))

print("\nConcatenated Array:")
print(concatenated)

# 13. Transformation
transformed = data * 2 + 5

print("\nTransformed Multidimensional Array:")
print(transformed)
```
