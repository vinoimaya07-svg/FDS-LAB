```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.array([
    12, 15, 18, 20, 22, 25, 27, 28, 30, 32,
    35, 36, 38, 40, 42, 43, 45, 47, 48, 50,
    52, 54, 55, 57, 58, 60, 62, 65, 68, 70,
    72, 75, 78, 80, 82, 85, 88, 90, 92, 95
])
```

```python
plt.figure(figsize=(8, 5))
plt.hist(data, bins=8, edgecolor='black')
plt.title("Histogram of Numerical Data")
plt.xlabel("Data Values")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
```

```python
bins = np.arange(10, 101, 10)
frequency, edges = np.histogram(data, bins=bins)
```

```python
print("=" * 55)
print("BINNED DISTRIBUTION OF NUMERICAL DATA")
print("=" * 55)
print("Bin Interval\tFrequency")
print("-" * 55)
for i in range(len(frequency)):
    print(f"{edges[i]:.0f} - {edges[i+1]:.0f}\t\t{frequency[i]}")
print("=" * 55)
```

```python
plt.figure(figsize=(8, 5))
plt.hist(data, bins=bins, edgecolor='black')
plt.title("Binned Distribution of Numerical Data")
plt.xlabel("Data Values")
plt.ylabel("Frequency")
plt.xticks(bins)
plt.grid(axis='y')
plt.show()
```
