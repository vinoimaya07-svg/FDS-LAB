```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# 1. CREATE DATASET

data = [
    45, 52, 48, 60, 55,
    47, 50, 62, 58, 51,
    49, 53, 57, 46, 54
]

# Create Pandas Series
series = pd.Series(data, name="Values")


# 2. BASIC STATISTICAL MEASURES

count = series.count()
total = series.sum()
minimum = series.min()
maximum = series.max()
data_range = maximum - minimum


# 3. MEASURES OF CENTRAL TENDENCY

mean = series.mean()
median = series.median()

mode_result = series.mode()
mode = mode_result.iloc[0]


# 4. MEASURES OF DISPERSION

variance = series.var()
standard_deviation = series.std()

mean_absolute_deviation = np.mean(
    np.abs(series - mean)
)

coefficient_of_variation = (
    standard_deviation / mean
) * 100


# 5. QUARTILES AND PERCENTILES

q1 = series.quantile(0.25)
q2 = series.quantile(0.50)
q3 = series.quantile(0.75)

interquartile_range = q3 - q1

percentile_10 = series.quantile(0.10)
percentile_90 = series.quantile(0.90)


# 6. SKEWNESS AND KURTOSIS

skewness = series.skew()
kurtosis = series.kurtosis()


# 7. OUTLIER DETECTION USING IQR METHOD

lower_bound = q1 - (1.5 * interquartile_range)
upper_bound = q3 + (1.5 * interquartile_range)

outliers = series[
    (series < lower_bound) |
    (series > upper_bound)
]


# 8. CORRELATION ANALYSIS

# Second related dataset
second_data = [
    43, 50, 47, 61, 54,
    46, 51, 63, 56, 52,
    48, 55, 58, 45, 53
]

second_series = pd.Series(
    second_data,
    name="Second Values"
)

correlation = series.corr(second_series)


# 9. DISPLAY DATA

print("=" * 70)
print("        STATISTICAL ANALYSIS USING DESCRIPTIVE MEASURES")
print("=" * 70)

print("\nDATASET")
print("-" * 70)

print(series.to_string(index=False))


# 10. DISPLAY BASIC STATISTICS

print("\nBASIC STATISTICAL MEASURES")
print("-" * 70)

print(f"Number of Observations : {count}")
print(f"Sum                    : {total:.2f}")
print(f"Minimum                : {minimum:.2f}")
print(f"Maximum                : {maximum:.2f}")
print(f"Range                  : {data_range:.2f}")


# 11. DISPLAY CENTRAL TENDENCY

print("\nMEASURES OF CENTRAL TENDENCY")
print("-" * 70)

print(f"Mean                   : {mean:.2f}")
print(f"Median                 : {median:.2f}")
print(f"Mode                   : {mode:.2f}")


# 12. DISPLAY DISPERSION

print("\nMEASURES OF DISPERSION")
print("-" * 70)

print(f"Variance               : {variance:.2f}")
print(f"Standard Deviation     : {standard_deviation:.2f}")
print(f"Mean Absolute Deviation: {mean_absolute_deviation:.2f}")
print(f"Coefficient of Variation (%): {coefficient_of_variation:.2f}")


# 13. DISPLAY QUARTILES

print("\nQUARTILES AND PERCENTILES")
print("-" * 70)

print(f"First Quartile (Q1)    : {q1:.2f}")
print(f"Second Quartile (Q2)   : {q2:.2f}")
print(f"Third Quartile (Q3)    : {q3:.2f}")
print(f"Interquartile Range    : {interquartile_range:.2f}")
print(f"10th Percentile        : {percentile_10:.2f}")
print(f"90th Percentile        : {percentile_90:.2f}")


# 14. DISPLAY SKEWNESS AND KURTOSIS

print("\nSHAPE OF DISTRIBUTION")
print("-" * 70)

print(f"Skewness               : {skewness:.4f}")
print(f"Kurtosis               : {kurtosis:.4f}")


# 15. DISPLAY OUTLIER INFORMATION

print("\nOUTLIER ANALYSIS USING IQR")
print("-" * 70)

print(f"Lower Bound            : {lower_bound:.2f}")
print(f"Upper Bound            : {upper_bound:.2f}")

if len(outliers) == 0:
    print("Outliers               : No outliers detected")
else:
    print("Outliers               :", outliers.tolist())


# 16. DISPLAY CORRELATION

print("\nCORRELATION ANALYSIS")
print("-" * 70)

print(f"Correlation Coefficient: {correlation:.4f}")


# 17. SUMMARY TABLE

summary = pd.DataFrame({

    "Measure": [
        "Count",
        "Sum",
        "Minimum",
        "Maximum",
        "Range",
        "Mean",
        "Median",
        "Mode",
        "Variance",
        "Standard Deviation",
        "Mean Absolute Deviation",
        "Coefficient of Variation (%)",
        "Q1",
        "Q2 (Median)",
        "Q3",
        "Interquartile Range",
        "10th Percentile",
        "90th Percentile",
        "Skewness",
        "Kurtosis"
    ],

    "Value": [
        count,
        total,
        minimum,
        maximum,
        data_range,
        mean,
        median,
        mode,
        variance,
        standard_deviation,
        mean_absolute_deviation,
        coefficient_of_variation,
        q1,
        q2,
        q3,
        interquartile_range,
        percentile_10,
        percentile_90,
        skewness,
        kurtosis
    ]
})

print("\nSUMMARY OF DESCRIPTIVE STATISTICS")
print("-" * 70)

print(summary.to_string(index=False))


# 18. HISTOGRAM

plt.figure(figsize=(12, 7))

plt.hist(
    series,
    bins=6,
    edgecolor="black"
)

plt.axvline(
    mean,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean:.2f}"
)

plt.axvline(
    median,
    linestyle="-.",
    linewidth=2,
    label=f"Median = {median:.2f}"
)

plt.title(
    "Distribution of Data",
    fontsize=18
)

plt.xlabel(
    "Data Values",
    fontsize=14
)

plt.ylabel(
    "Frequency",
    fontsize=14
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# 19. BOX PLOT

plt.figure(figsize=(12, 7))

plt.boxplot(
    series,
    vert=True,
    patch_artist=True
)

plt.title(
    "Box Plot of Dataset",
    fontsize=18
)

plt.ylabel(
    "Data Values",
    fontsize=14
)

plt.xticks(
    [1],
    ["Dataset"]
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# 20. COMPARISON OF CENTRAL TENDENCY MEASURES

measures = ["Mean", "Median", "Mode"]
values = [mean, median, mode]

plt.figure(figsize=(12, 7))

bars = plt.bar(
    measures,
    values,
    edgecolor="black"
)

plt.title(
    "Comparison of Descriptive Measures",
    fontsize=18
)

plt.xlabel(
    "Descriptive Measure",
    fontsize=14
)

plt.ylabel(
    "Value",
    fontsize=14
)

plt.grid(
    axis="y",
    alpha=0.3
)

for bar, value in zip(bars, values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.2,
        f"{value:.2f}",
        ha="center",
        fontsize=12
    )

plt.tight_layout()
plt.show()


print("\n" + "=" * 70)
print("Descriptive statistical analysis completed successfully.")
print("=" * 70)
```
