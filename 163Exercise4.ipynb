```python
import pandas as pd

# Create student DataFrame
data = {
    'Student ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Name': ['Arun', 'Bala', 'Charan', 'Divya', 'Esha', 'Farhan', 'Gokul', 'Hema'],
    'Department': ['CSE', 'ECE', 'CSE', 'IT', 'ECE', 'CSE', 'IT', 'ECE'],
    'Gender': ['Male', 'Male', 'Male', 'Female', 'Female', 'Male', 'Male', 'Female'],
    'Attendance': [85, 72, 90, 65, 88, 78, 92, 80],
    'Internal Marks': [42, 35, 45, 30, 40, 38, 44, 36],
    'External Marks': [45, 40, 48, 35, 42, 39, 46, 41],
    'Study Hours': [5, 3, 6, 2, 5, 4, 6, 4]
}

df = pd.DataFrame(data)

print("Student DataFrame:")
display(df)
```

```python
print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataFrame Information:")
df.info()
```

```python
# Select a single column
print("Student Names:")
print(df['Name'])

# Select multiple columns
print("\nName and Department:")
print(df[['Name', 'Department']])

# Using loc[]
print("\nUsing loc[]:")
print(df.loc[0:3, ['Name', 'Department', 'Attendance']])

# Using iloc[]
print("\nUsing iloc[]:")
print(df.iloc[0:4, 0:4])
```

```python
# Students with attendance >= 80
high_attendance = df[df['Attendance'] >= 80]

print("Students with Attendance >= 80:")
print(high_attendance[['Name', 'Attendance']])

# Students with Internal Marks >= 40
high_internal = df[df['Internal Marks'] >= 40]

print("\nStudents with Internal Marks >= 40:")
print(high_internal[['Name', 'Internal Marks']])

# Students with External Marks >= 80
high_external = df[df['External Marks'] >= 80]

print("\nStudents with External Marks >= 80:")
print(high_external[['Name', 'External Marks']])
```

```python
df['Total Marks'] = df['Internal Marks'] + df['External Marks']

df['Percentage'] = (df['Total Marks'] / 150) * 100

print(df[['Name', 'Internal Marks', 'External Marks',
          'Total Marks', 'Percentage']])
```

```python
def performance_category(percentage):
    if percentage >= 85:
        return 'Excellent'
    elif percentage >= 70:
        return 'Good'
    elif percentage >= 60:
        return 'Average'
    else:
        return 'Needs Improvement'

df['Performance'] = df['Percentage'].apply(performance_category)

print(df[['Name', 'Percentage', 'Performance']])
```

```python
# Sort by percentage in descending order
sorted_percentage = df.sort_values(by='Percentage', ascending=False)

print("Sorted according to Percentage:")
print(sorted_percentage[['Name', 'Percentage']])

# Sort by attendance in descending order
sorted_attendance = df.sort_values(by='Attendance', ascending=False)

print("\nSorted according to Attendance:")
print(sorted_attendance[['Name', 'Attendance']])
```

```python
department_group = df.groupby('Department').agg({
    'Percentage': ['mean', 'max', 'min'],
    'Attendance': ['mean', 'max', 'min'],
    'Student ID': 'count'
})

print(department_group)
```

```python
dept_gender_group = df.groupby(['Department', 'Gender']).agg({
    'Percentage': 'mean',
    'Attendance': 'mean',
    'Student ID': 'count'
})

print(dept_gender_group)
```

```python
department_data = {
    'Department': ['CSE', 'ECE', 'IT'],
    'HOD': ['Dr. Kumar', 'Dr. Priya', 'Dr. Ravi'],
    'Room No': [101, 102, 103]
}

department_df = pd.DataFrame(department_data)

print("Department DataFrame:")
print(department_df)

# Merge with student DataFrame
merged_df = pd.merge(df, department_df, on='Department', how='left')

print("\nMerged DataFrame:")
print(merged_df)
```

```python
pivot_table = pd.pivot_table(
    df,
    values='Percentage',
    index='Department',
    columns='Gender',
    aggfunc='mean'
)

print("Pivot Table:")
print(pivot_table)
```
